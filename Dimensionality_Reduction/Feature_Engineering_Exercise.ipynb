## 1. Imports

```python
# Imports
import numpy as np
import pandas as pd

pd.set_option('display.max_columns',100)
```

##2. Load the Data

```python
df = pd.read_csv("/content/drive/MyDrive/AXSOSACADEMY/03-DataEnrichment/Week09/Data/bikeshare_train - bikeshare_train.csv")
df.info()
df.head()
```

##3. Drop Redundant Columns

```python
df.drop(columns=['casual', 'registered'], inplace=True)
```

##4. Convert Datetime Column

```python
df['datetime'] = pd.to_datetime(df['datetime'])
```

##5. Create New Datetime Features

```python
df['month'] = df['datetime'].dt.month_name()
df['day'] = df['datetime'].dt.day_name()
df['hour'] = df['datetime'].dt.hour
```

```python
df.head()
```

##6. Convert New Columns to Object Type

```python
df.dtypes
```

```python
df['hour'] = df['hour'].astype('object')
```

```python
df.dtypes
```

##7. Drop Redundant Columns

```python
df.drop(columns=['datetime', 'season'], inplace=True)
```

##8. Convert Celsius to Fahrenheit Using apply + lambda

```python
df['temp'] = df['temp'].apply(
    lambda x: (x * 9/5) + 32
)
```

```python
df['atemp'] = df['atemp'].apply(
    lambda x: (x * 9/5) + 32
)
```

##9. Create temp_variance Feature

```python
df['temp_variance'] = df['temp'] - df['atemp']
```

```python
df.drop(columns='atemp', inplace=True)
```

```python
df.head()
```

# Feature Engineering Summary

Several new features were engineered from the original dataset to improve the model’s ability to predict hourly bike-share rentals.

The `datetime` column was converted into a datetime datatype, and new features were extracted from it, including:
- Month name
- Day name
- Hour of the day

These features may help the model detect seasonal and hourly rental patterns.

The temperature-related columns (`temp` and `atemp`) were converted from Celsius to Fahrenheit using `.apply()` with lambda functions.

A new feature called `temp_variance` was also created to measure how much warmer or colder the actual temperature was compared to the “feels like” temperature.

Finally, redundant columns such as `datetime`, `season`, and `atemp` were removed.
