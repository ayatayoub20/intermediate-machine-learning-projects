```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from scipy.spatial.distance import cdist
```

#1. Load Data

```python
df = pd.read_csv('/content/drive/MyDrive/AXSOSACADEMY/03-DataEnrichment/Week10/Data/credit_card.csv')
df.head()
```

```python
df.info()
df.isna().sum()
```

#2. Prepare X

```python
X = df.copy()
```

#3. Scale Data

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

#4. Fit KMeans with 3 clusters

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init='auto')
kmeans.fit(X_scaled)
```

#5. Calculate distances using cdist

```python
distances = cdist(X_scaled, kmeans.cluster_centers_, metric='euclidean')
distances.shape
```

#6. Get minimum distance for each transaction

```python
min_distances = np.min(distances, axis=1)
min_distances
```

#7. Define threshold using 99.6 percentile

```python
threshold = float(np.percentile(min_distances, 99.6))
threshold
```

#8. Get anomalous transactions indices

```python
anomaly_filter = min_distances > threshold

anomaly_indices = X[anomaly_filter].index.tolist()

len(anomaly_indices)
```

#9. View anomalous transactions

```python
anomalies = df.loc[anomaly_indices]

anomalies.head()
```

#10. Add results to original dataframe

```python
df['KMeans_Min_Distance'] = min_distances
df['KMeans_Anomaly'] = anomaly_filter.astype(int)

df[['KMeans_Min_Distance', 'KMeans_Anomaly']].head()
```

## KMeans Anomaly Detection

For this task, I used KMeans clustering to detect anomalous credit card transactions. Since the stakeholder reported that approximately 99.6% of transactions are valid and 0.4% are fraudulent, I used the 99.6th percentile of the minimum distances to the cluster centers as the anomaly threshold.

First, I scaled the data because KMeans is distance-based. Then, I fitted a KMeans model with 3 clusters using a random state of 42. After fitting the model, I calculated the Euclidean distance between each transaction and each cluster center using `cdist`.

For each transaction, I kept the minimum distance to any cluster center. Transactions with a minimum distance greater than the 99.6th percentile threshold were identified as anomalous.
